```python
from pathlib import Path
import shutil

original = Path("/kaggle/input/notebooks/purveee04/data-preparation-final/gc_swgan_fixed.py")
patched = Path("/kaggle/working/gc_swgan_fixed.py")

shutil.copy(original, patched)

code = patched.read_text()

old = '''assert gen.output_shape[1:] == (args.img_size, args.img_size, CH), gen.output_shape
print("G params: %d | shared trunk params: %d" % (gen.count_params(), trunk.count_params()))
'''

new = '''assert gen.output_shape[1:] == (args.img_size, args.img_size, CH), gen.output_shape

# Build classifier and critic before accessing their trainable variables.
dummy_input = tf.zeros((1, args.img_size, args.img_size, CH))
_ = clf(dummy_input)
_ = critic(dummy_input)

print("G params: %d | shared trunk params: %d" % (gen.count_params(), trunk.count_params()))
'''

if old not in code:
    raise ValueError("Expected code section not found. Original script was not modified.")

patched.write_text(code.replace(old, new, 1))

print("Patched copy created:", patched)
print("Original script remains unchanged.")
```

```python
from pathlib import Path

script = Path("/kaggle/working/gc_swgan_fixed.py")
code = script.read_text()

# 1. Add 128 to the supported resolutions
code = code.replace(
    'N_UP = {192: 5, 96: 4, 48: 3}[args.img_size]',
    'N_UP = {192: 5, 128: 5, 96: 4, 48: 3}[args.img_size]'
)

# 2. Use the existing 5-stage architecture for 128
# No change is needed to G_HIDDEN or D_CFG.

# 3. Make the Generator's initial feature map resolution-dependent
code = code.replace(
    'm.add(layers.Dense(6 * 6 * 256, use_bias=False))',
    'base_size = args.img_size // (2 ** N_UP)\\n    m.add(layers.Dense(base_size * base_size * 256, use_bias=False))'
)

code = code.replace(
    'm.add(layers.Reshape((6, 6, 256)))',
    'm.add(layers.Reshape((base_size, base_size, 256)))'
)

script.write_text(code)

print("128x128 support added to the working script.")
```
